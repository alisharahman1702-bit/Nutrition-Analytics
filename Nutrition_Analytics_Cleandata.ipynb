## Section 6 — Data Cleaning
We keep a copy of the raw data first, then clean **only what is genuinely needed**.
Each step explains what it does and why.

```python
df_original = df.copy()          # raw copy for the before/after summary
rows_before = df.shape[0]
cols_before = df.shape[1]
```

### A. Text Cleaning
Remove leading/trailing spaces and collapse double spaces in the text columns so
the same category is not accidentally split into two.

```python
text_cols = df.select_dtypes(include="object").columns.tolist()
for c in text_cols:
    df[c] = df[c].astype(str).str.strip().str.replace(r"\s+", " ", regex=True)

# check whether any inconsistent spacing remained
left = sum((df[c] != df[c].str.strip()).sum() for c in text_cols)
print("Text columns cleaned:", text_cols)
print("Remaining spacing issues:", left)
print("Note: capitalization was NOT forced to lower/upper — food names are proper "
      "labels and are kept in their original readable form.")
```

### B. Numerical Data Cleaning
Make sure every numeric column really is numeric. Anything non-numeric becomes NaN
so it can be handled with the other missing values.
*(In this dataset the types were already correct — this step just makes the code safe.)*

```python
expected_numeric = ['Serving Size (g)', 'Calories (kcal)', 'Protein (g)',
                    'Total Fat (g)', 'Carbohydrates (g)', 'Fiber (g)', 'Sugar (g)',
                    'Sodium (mg)', 'Cholesterol (mg)', 'Potassium (mg)']
numeric_cols = [c for c in expected_numeric if c in df.columns]   # safe: only existing ones

for c in numeric_cols:
    df[c] = pd.to_numeric(df[c], errors="coerce")
print("Confirmed numeric columns:", numeric_cols)
```

### C. Impossible / Inconsistent Values
All values are per **100 g** of food, so some values are physically impossible:
* a solid nutrient (protein, fat, carbs, fiber, sugar) cannot exceed **100 g** per 100 g, and
* calories cannot exceed about **900 kcal** per 100 g (pure fat is the maximum).

Rows breaking these physical limits are scraping errors, so we remove them. We do
**not** remove merely large-but-valid values (e.g. oils really do have ~800 kcal).

```python
gram_cols = [c for c in ['Protein (g)', 'Total Fat (g)', 'Carbohydrates (g)',
                         'Fiber (g)', 'Sugar (g)'] if c in df.columns]

impossible = pd.Series(False, index=df.index)
if 'Calories (kcal)' in df.columns:
    impossible |= df['Calories (kcal)'] > 900
if gram_cols:
    impossible |= (df[gram_cols] > 100).any(axis=1)

print("Impossible rows found:", int(impossible.sum()))
if impossible.sum() > 0:
    df = df[~impossible].reset_index(drop=True)
    print("Removed. Rows now:", len(df))
else:
    print("No impossible values found.")

# also check for negative values (impossible for nutrition)
neg = (df[numeric_cols] < 0).any().any()
print("Negative values present:", neg, "(none expected)")
```

### D. Handle Missing Values (per-column reasoning)
We choose the method based on what each column **means** — not a blanket 0:

* **Core nutrients** (Calories, Protein, Total Fat, Carbohydrates): essential, and
  only a few rows miss them → **drop** those rows (keeps the data real).
* **Fiber, Sugar, Sodium, Cholesterol**: on a nutrition label a nutrient that is
  not listed is absent/negligible, so **0** is the correct value here (using the
  median would invent content the food does not have).
* **Potassium**: about two-thirds is missing, so dropping would destroy the data
  and 0 would be wrong (foods do contain potassium). We fill each gap with the
  **median of its food category** — a neutral estimate. Because so much of this
  column is estimated, we treat Potassium results with caution later.

```python
# 1) drop rows missing a core nutrient
core = [c for c in ['Calories (kcal)', 'Protein (g)', 'Total Fat (g)',
                    'Carbohydrates (g)'] if c in df.columns]
before = len(df)
df = df.dropna(subset=core).reset_index(drop=True)
print("Rows dropped for missing core nutrient:", before - len(df))

# 2) label-absent nutrients -> 0
for c in ['Fiber (g)', 'Sugar (g)', 'Sodium (mg)', 'Cholesterol (mg)']:
    if c in df.columns:
        df[c] = df[c].fillna(0)

# 3) potassium -> category median, then global median for any leftover
if 'Potassium (mg)' in df.columns:
    gmed = df['Potassium (mg)'].median()
    df['Potassium (mg)'] = (df.groupby('Food Category')['Potassium (mg)']
                              .transform(lambda s: s.fillna(s.median())))
    df['Potassium (mg)'] = df['Potassium (mg)'].fillna(gmed)

print("Missing values remaining:", int(df.isnull().sum().sum()))
```

### E. Remove Unnecessary Column (only if genuinely useless)
`Serving Size (g)` holds the **same value (100) in every row**, so it carries no
information for analysis. We drop it only after confirming it is constant.

```python
if 'Serving Size (g)' in df.columns and df['Serving Size (g)'].nunique() == 1:
    print("'Serving Size (g)' is constant =", df['Serving Size (g)'].unique()[0],
          "-> removing it (no analytical value).")
    df = df.drop(columns=['Serving Size (g)'])
else:
    print("'Serving Size (g)' varies, so it is kept.")

numeric_cols = [c for c in numeric_cols if c in df.columns]   # refresh list
print("Columns now:", list(df.columns))
```

### F. Outlier Detection (detect, do NOT auto-delete)
We use the **IQR method** to *count* outliers in the main nutrients. These are
mostly legitimate foods (e.g. oils high in fat, dried fruit high in sugar), so we
**keep** them — the only outliers we already removed were the physically impossible
ones in step C.

```python
def count_iqr_outliers(series):
    q1, q3 = series.quantile(0.25), series.quantile(0.75)
    iqr = q3 - q1
    low, high = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    return int(((series < low) | (series > high)).sum())

print("Outliers detected by the IQR rule (kept, because they are valid foods):")
for c in numeric_cols:
    print(f"  {c}: {count_iqr_outliers(df[c])} outliers")
```

### Cleaning steps that were NOT needed
* **Duplicate removal** — 0 exact duplicates existed.
* **Negative-value fixing** — no negative values existed.
* **Category spelling/space fixes** — none remained after trimming spaces.
* **Outlier deletion** — the remaining outliers are valid foods, so nothing was deleted here.

## Section 8 — Data Analysis
All numbers below are computed directly from the cleaned data.

### A. Numerical Analysis — mean, median, min, max, std

```python
summary = df[numeric_cols].agg(['mean', 'median', 'min', 'max', 'std']).round(2)
display(summary)
```

```python
# highest and lowest food for a few key nutrients
for c in ['Calories (kcal)', 'Protein (g)', 'Sugar (g)']:
    if c in df.columns:
        hi, lo = df.loc[df[c].idxmax()], df.loc[df[c].idxmin()]
        print(f"{c}")
        print(f"   Highest: {hi['Food Name'][:55]} -> {hi[c]}")
        print(f"   Lowest : {lo['Food Name'][:55]} -> {lo[c]}\n")
```

### B. Categorical Analysis — counts, most/least common

```python
if 'Food Category' in df.columns:
    counts = df['Food Category'].value_counts()
    print("Total categories     :", df['Food Category'].nunique())
    print("Most common category :", counts.idxmax(), "->", counts.max(), "foods")
    print("Least common category:", counts.idxmin(), "->", counts.min(), "food(s)")
    print("\nTop 10 categories by number of foods:")
    display(counts.head(10))
else:
    print("No categorical column available, so categorical analysis is skipped.")
```

### C. Group-Based Analysis — average nutrients per category

```python
# use only categories with >=10 foods so the averages are reliable
group = df.groupby('Food Category').agg(
    Foods=('Food Name', 'count'),
    Avg_Calories=('Calories (kcal)', 'mean'),
    Avg_Protein=('Protein (g)', 'mean'),
    Avg_Sugar=('Sugar (g)', 'mean')).round(1)
group_reliable = group[group['Foods'] >= 10]

print("Top 10 categories by AVERAGE CALORIES (>=10 foods):")
display(group_reliable.sort_values('Avg_Calories', ascending=False).head(10))
print("Top 10 categories by AVERAGE PROTEIN (>=10 foods):")
display(group_reliable.sort_values('Avg_Protein', ascending=False).head(10))
```

### D. Relationship Analysis — correlation
We only treat a relationship as strong when the correlation is at least 0.6 in size.

```python
if len(numeric_cols) >= 2:
    corr = df[numeric_cols].corr().round(2)
    display(corr)

    print("Strong relationships (|correlation| >= 0.6):")
    found = False
    for i in range(len(numeric_cols)):
        for j in range(i + 1, len(numeric_cols)):
            r = corr.iloc[i, j]
            if abs(r) >= 0.6:
                found = True
                kind = "positive" if r > 0 else "negative"
                print(f"  {numeric_cols[i]}  &  {numeric_cols[j]}: {r} ({kind})")
    if not found:
        print("  none")
else:
    print("Fewer than 2 numeric columns, so correlation analysis is skipped.")
```
